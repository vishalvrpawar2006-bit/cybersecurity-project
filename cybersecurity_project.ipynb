## Module 1: Password Strength & Entropy Analyzer\nThis module evaluates password complexity and calculates Shannon entropy to gauge resistance against brute-force attacks.

```python
def calculate_entropy(password):\n    if not password:\n        return 0.0\n    \n    pool_size = 0\n    if re.search(r'[a-z]', password): pool_size += 26\n    if re.search(r'[A-Z]', password): pool_size += 26\n    if re.search(r'[0-9]', password): pool_size += 10\n    if re.search(r'[^a-zA-Z0-9]', password): pool_size += 32\n    \n    if pool_size == 0:\n        return 0.0\n        \n    entropy = len(password) * math.log2(pool_size)\n    return round(entropy, 2)\n\ndef evaluate_password_strength(password):\n    score = 0\n    feedback = []\n    \n    if len(password) >= 12:\n        score += 2\n    elif len(password) >= 8:\n        score += 1\n        feedback.append(\"Consider increasing password length to 12+ characters.\")\n    else:\n        feedback.append(\"Password is too short (less than 8 characters).\")\n        \n    if re.search(r'[A-Z]', password) and re.search(r'[a-z]', password):\n        score += 1\n    else:\n        feedback.append(\"Include both uppercase and lowercase letters.\")\n        \n    if re.search(r'[0-9]', password):\n        score += 1\n    else:\n        feedback.append(\"Include numbers (0-9).\")\n        \n    if re.search(r'[^a-zA-Z0-9]', password):\n        score += 1\n    else:\n        feedback.append(\"Include special symbols (!@#$%^&*).\")\n        \n    entropy = calculate_entropy(password)\n    \n    if score >= 5 and entropy > 60:\n        rating = \"Strong\"\n    elif score >= 3 and entropy > 35:\n        rating = \"Moderate\"\n    else:\n        rating = \"Weak\"\n        \n    return {\n        \"rating\": rating,\n        \"score\": score,\n        \"entropy_bits\": entropy,\n        \"feedback\": feedback\n    }\n\ntest_password = \"SecurePassword@2026!\"\nresult = evaluate_password_strength(test_password)\nprint(f\"Password Evaluated: {test_password}\")\nprint(f\"Rating: {result['rating']}\")\nprint(f\"Entropy: {result['entropy_bits']} bits\")\nprint(f\"Feedback: {result['feedback']}\")
```
